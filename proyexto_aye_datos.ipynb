## B. Outliers

El método del rango intercuartílico (IQR, por sus siglas en inglés) es una técnica estadística robusta para detectar valores atípicos (outliers). Con él se pueden identificar puntos de datos inusualmente altos o bajos calculando el 50% central de los datos y estableciendo límites más allá de los cuales las observaciones puedan considerse extremos.

```python
# Detección de outliers usando el método IQR (Rango Intercuartílico)
for col in ['Quantity', 'Price', 'Total']:
    Q1 = df_limpio[col].quantile(0.25)
    Q3 = df_limpio[col].quantile(0.75)
    IQR = Q3 - Q1
    limite_inf = Q1 - 1.5 * IQR
    limite_sup = Q3 + 1.5 * IQR

    outliers = df_limpio[(df_limpio[col] < limite_inf) | (df_limpio[col] > limite_sup)]
    pct = (len(outliers) / len(df_limpio)) * 100

    print(f"--- {col} ---")
    print(f"  Límite inferior: {limite_inf:.2f}")
    print(f"  Límite superior: {limite_sup:.2f}")
    print(f"  Outliers detectados: {len(outliers)} ({pct:.2f}% del total)")
    print()
```

###  Conclusiones sobre Outliers

Se aplicó el método IQR para detectar valores extremos en las tres variables numéricas clave:

- Quantity: Se detectaron 51,734 outliers (5.26% del total), correspondientes a transacciones
  con más de 28.50 unidades. El límite inferior dio -15.50, lo que en la práctica no aplica
  ya que tras la limpieza no hay cantidades negativas en df_limpio (el cual contiene únicamente las transacciones de venta).

- Price: Se detectaron 77,802 outliers (7.84% del total), es decir, productos con precio
  unitario superior a £8.45. Considerando que la mayoría de los productos cuestan menos de
  £2.10 (mediana), precios por encima de ese umbral ya son considerados atípicos por el método,
  aunque en la realidad son perfectamente válidos.

- Total: Se detectaron 82,835 outliers (8.42% del total), transacciones que superan los
  £38.06 de ingreso por línea. Nuevamente, esto refleja la asimetría del dataset más que
  errores reales.

Decisión: se decide NO eliminar los outliers por las siguientes razones:

1. Este es un dataset de retail mayorista, donde es completamente normal que ciertos clientes
   compren volúmenes muy grandes o productos de alto precio. Eliminarlos distorsionaría

## C. Correlación

La correlación  es una medida estadística que cuantifica el grado y la dirección en la que dos variables cuantitativas se modifican al mismo tiempo. Permite entender si, a medida que el valor de una variable aumenta o disminuye, la otra variable tiende a seguir un patrón predecible. La relación se expresa mediante un valor numérico llamado coeficiente de correlación (generalmente representado con la letra $r$), el cual siempre oscila entre $-1$ y $+1$.

- Correlación positiva ($r > 0$): Si una variable aumenta, la otra también lo hace. Una correlación de $1$ indica una relación positiva perfecta.

- Correlación negativa ($r < 0$): Si una variable aumenta, la otra disminuye. Una correlación de $-1$ indica una relación negativa perfecta.

- Correlación nula ($r = 0$): No existe relación lineal entre las variables; el cambio en una no nos dice nada sobre el comportamiento de la otra.

Se utilizará el método de Pearson para cuantificar la correlación. Mide la relación lineal entre dos variables continuas. Asume que los datos tienen una distribución normal y es muy sensible a los valores atípicos. En la librería Pandas está representado por el método **.corr()** con sus parámetros por defecto.

```python
# Correlación entre variables numéricas
correlacion = df_limpio[['Quantity', 'Price', 'Total']].corr()
print(correlacion)
```

### Análisis sobre la Correlación

- Quantity vs Total (0.81): Es la correlación más fuerte del análisis. Tiene sentido:
  a mayor cantidad de unidades compradas, mayor es el ingreso total de esa línea. El valor
  de 0.81 indica una relación lineal positiva muy fuerte, lo que confirma que el volumen
  de unidades es el principal motor del ingreso por transacción.

- Price vs Total (0.25): Correlación positiva pero débil. Aunque el precio unitario
  contribuye al total, su impacto es mucho menor que el de la cantidad. Esto sugiere que
  los clientes que más ingresos generan no son necesariamente los que compran productos
  caros, sino los que compran en mayor volumen.

- Quantity vs Price (-0.004): Correlación prácticamente nula. No existe ninguna relación
  lineal entre el precio de un producto y la cantidad en que se compra. Esto indica que
  los clientes no ajustan su volumen de compra en función del precio unitario, algo
  coherente con un contexto mayorista donde las decisiones de compra dependen más
  de la demanda del producto que de su precio.

Conclusión general: el volumen de unidades (Quantity) es la variable que más
explica los ingresos por transacción. El precio unitario tiene un rol secundario.
Desde una perspectiva comercial, las estrategias orientadas a aumentar el volumen
por pedido (descuentos por cantidad, promociones mayoristas) tendrían mayor impacto
en los ingresos que subir precios.
